**Este código importa y estandariza las variables de *unidades.xlsx***

```python
import pandas as pd
import pathlib
import janitor
import unidecode
```

La siguiente celda de código contiene líneas que separan las rutas relativas de la raíz. Esto es epecialmente útil en casos de trabajo colaborativo, ya que solo es necesario modificar una vez la ruta del objeto llamado *here*. Por ejemplo, si varias personas estuvieran trabajando en un repositorio que contiene este código, modificarían la ruta de here por la ruta en la que tienen clonado el repositorio en su máquina local, y el resto de rutas permanecerían igual.

```python
# Setup

here = pathlib.Path("/Users/macbookair/Documents/prueba-micronegocios/importar").resolve().parent

unidades_path = here/'importar/input/unidades.xlsx'
output = here/'importar/output/unidades.xlsx'
```

```python
# Leer datos 

unidades = pd.read_excel(unidades_path).clean_names()
unidades.head()
```

```python
# Estandarización: mayúsculas y tildes

unidades = unidades.map(lambda x: unidecode.unidecode(x).lower() if isinstance(x, str) else x)
```

```python
unidades.head()
```

```python
# Save 

unidades.to_excel(output, index=False)
```
