**Este Notebook desarrolla los puntos de la prueba de datos de la SDEE con las bases de datos estandarizadas**

```python
import pandas as pd
import pathlib
from IPython.display import display, Markdown
import seaborn as sns
import matplotlib.pyplot as plt
import re
```

```python
# Setup

# Esta celda contiene líneas de código que separan las rutas relativas de la raíz del directorio. 
# Esto es epecialmente útil en casos de trabajo colaborativo, ya que solo es necesario modificar 
# una vez la ruta del objeto llamado here. 
# Por ejemplo, si varias personas estuvieran trabajando en un repositorio que contiene este código, 
# modificarían la ruta de here por la ruta en la que tienen clonado el repositorio en su máquina local,
# y el resto de rutas, relativas al repositorio, permanecerían igual. 

here = pathlib.Path("/Users/macbookair/Documents/prueba-micronegocios/analisis").resolve().parent

desembolsos_path = here/'importar/output/desembolsos.xlsx'
unidades_path = here/'importar/output/unidades.xlsx'
parametros_path = here/'importar/input/paramétricas.xlsx'
```

```python
# Leer datos 

unidades = pd.read_excel(unidades_path)
desembolsos = pd.read_excel(desembolsos_path)
parametros = pd.read_excel(parametros_path)
localidades = pd.read_excel(parametros_path, sheet_name=1)
```

```python
# Cruce

joint = pd.merge(unidades, desembolsos, on = 'id_micronegocio', how = 'outer')
```

**Punto 1.A**

A continuación se hace un análisis del promedio de los desembolsos recibidos por **sexo  y grupo étnico** del propietario, **localidad** en la que se encuentea el negocio y **sector económico**. 

```python
# Descriptivas por sexo

sexo = joint.groupby('sexo').agg(total=('desembolso', 'sum'),
                                 media=('desembolso', 'mean'),
                                numero_negocios = ('sexo', 'size')).reset_index()
sexo['total'] = sexo['total'].apply(lambda x: f'{x:.2f}')
sexo['media'] = sexo['media'].apply(lambda x: f'{x:.2f}')

sexo = sexo.to_markdown(index=False, floatfmt=".2f")
titulo = "Descritpivas por sexo"
display(Markdown(titulo))
display(Markdown(sexo))
```

```python
# Descriptivas por etnia

etnia = joint.groupby('grupo_etnico').agg(total=('desembolso', 'sum'),
                                          media=('desembolso', 'mean'),
                                         numero_negocios = ('grupo_etnico', 'size')).reset_index()
etnia['total'] = etnia['total'].apply(lambda x: f'{x:.2f}')
etnia['media'] = etnia['media'].apply(lambda x: f'{x:.2f}')

etnia = etnia.to_markdown(index=False, floatfmt=".2f")
titulo = "Descritpivas por etnia"
display(Markdown(titulo))
display(Markdown(etnia))
```

```python
# Descriptivas por sexo y etnia

etnia_sexo = joint.groupby(['sexo', 'grupo_etnico']).agg(total=('desembolso', 'sum'),
                                                        media=('desembolso', 'mean'), 
                                                        numero_negocios = ('sexo', 'size')).reset_index()
etnia_sexo['total'] = etnia_sexo['total'].apply(lambda x: f'{x:.2f}')
etnia_sexo['media'] = etnia_sexo['media'].apply(lambda x: f'{x:.2f}')

etnia_sexo = etnia_sexo.to_markdown(index=False, floatfmt=".2f")
titulo = "Descritpivas por sexo y etnia"
display(Markdown(titulo))
display(Markdown(etnia_sexo))
```

```python
# Descriptivas por localidad

localidad = joint.groupby('localidad').agg(total=('desembolso', 'sum'),
                                           media=('desembolso', 'mean'), 
                                           numero_negocios = ('localidad', 'size')).reset_index()
localidad['total'] = localidad['total'].apply(lambda x: f'{x:.2f}')
localidad['media'] = localidad['media'].apply(lambda x: f'{x:.2f}')

localidad = localidad.to_markdown(index=False, floatfmt=".2f")
titulo = "Descritpivas por localidad"
display(Markdown(titulo))
display(Markdown(localidad))
```

```python
# Descriptivas por sector económico

sector = joint.groupby('sector_economico').agg(total=('desembolso', 'sum'),
                                               media=('desembolso', 'mean'),
                                               numero_negocios = ('sector_economico', 'size')).reset_index()
sector['total'] = sector['total'].apply(lambda x: f'{x:.2f}')
sector['media'] = sector['media'].apply(lambda x: f'{x:.2f}')

sector = sector.to_markdown(index=False, floatfmt=".2f")
titulo = "Descritpivas por sector económico"
display(Markdown(titulo))
display(Markdown(sector))
```

**Punto 1.B**

A continuación se parametrizan los valores de distintas variables de acuerdo a las reglas especificadas. Para esto, se define una función que crea un diccionario de mapeo que identifica los valores iniciales y les asigna una regla de acuerdo al archivo parametricas.xlsx. 

```python
# Función que crea un diccionario de mapeo tomando como insumo el archivo parametricas

def crear_mapeos(parametros):
    mapeos = {}
    for _, row in parametros.iterrows():
        columna = row['Nombre Columna']
        valor_permitido = row['Valores permitidos']
        clave = valor_permitido.split('(')[0].strip().lower().replace('ó', 'o')  # Ajuste simplificado
        if columna not in mapeos:
            mapeos[columna] = {}
        mapeos[columna][fr'\b{clave}\b'] = valor_permitido
    return mapeos

mapeos = crear_mapeos(parametros)
```

Se define una función para recodificar los valores de acuerdo al diccionario de mapeo, se especifican las columnas a recodificar y se aplica la función. 

```python
# Función para recodificar usando expresiones regulares

def reemplazar_con_regex(columna, mapeo):
    columna = columna.fillna('').astype(str)
    for patron, reemplazo in mapeo.items():
        columna = columna.apply(lambda x: re.sub(patron, reemplazo, x, flags=re.IGNORECASE))
    return columna
```

```python
# Columnas a recodificar y aplicación de la función

mapeo_columnas = {
    'estado_civil': 'EstadoCivil',
    'grupo_etnico': 'GrupoEtnico',
    'lugar_actividades': 'LugarDesarrollo',
    'medio_comercializacion': 'MedioComercializacion',
    'nivel_educativo': 'NivelEducativo',
    'sector_economico': 'SectorEconomicoSDDE',
    'sexo': 'SexoBiologico'
}

for columna_joint, columna_parametro in mapeo_columnas.items():
    if columna_joint in joint.columns:
        joint[columna_joint] = reemplazar_con_regex(joint[columna_joint], mapeos.get(columna_parametro, {}))
    
```

Modificación del formato de fechas de acuerdo a las reglas en *parametricas*

```python
# Recodificando fechas

joint['fecha_de_nacimiento'] = pd.to_datetime(joint['fecha_de_nacimiento'], format='%Y-%m-%d')
joint['fecha_de_nacimiento'] = joint['fecha_de_nacimiento'].dt.strftime('%d/%m/%Y')

joint['fecha_inicio_operaciones'] = pd.to_datetime(joint['fecha_inicio_operaciones'], format='%Y-%m-%d')
joint['fecha_inicio_operaciones'] = joint['fecha_inicio_operaciones'].dt.strftime('%d/%m/%Y')
```

Modificación de las localidades a sus códigos de acuerdo a las reglas en *parametricas*

```python
# Recodificando localidades

mapeo_localidad = {
    nombre.lower(): codigo
    for nombre, codigo in zip(localidades['NOMBRE LOCALIDAD'].str.lower(), localidades['CODIGO LOCALIDAD (valores permitidos)'])
}

joint['localidad'] = joint['localidad'].str.lower().map(mapeo_localidad)
```
